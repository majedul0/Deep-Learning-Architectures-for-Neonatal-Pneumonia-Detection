```python
import os, gc, random, json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, Model
from tensorflow.keras.applications import vgg16, resnet50, densenet, efficientnet, mobilenet_v2
from sklearn.metrics import (classification_report, confusion_matrix, roc_curve, auc,
                              precision_recall_curve, average_precision_score)

from google.colab import drive
drive.mount('/content/drive')

SEED = 42
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)
print("GPU available:", tf.config.list_physical_devices('GPU'))
```

Cell 2 — Config + copy dataset to local disk

```python
DRIVE_DATA_DIR = "/content/drive/MyDrive/Chest X-ray Pneumonia/Final_Dataset_Split"
DATA_DIR = "/content/Final_Dataset_Split"   # local copy -- fast, avoids Drive-mount stalls
TRAIN_DIR = os.path.join(DATA_DIR, "train")
VAL_DIR   = os.path.join(DATA_DIR, "val")
TEST_DIR  = os.path.join(DATA_DIR, "test")

RESULTS_DIR = "/content/drive/MyDrive/Chest X-ray Pneumonia/Results"  # results still saved to Drive
os.makedirs(RESULTS_DIR, exist_ok=True)

import shutil
if not os.path.exists(DATA_DIR):
    print("Copying dataset from Drive to local disk (one-time, a few minutes)...")
    shutil.copytree(DRIVE_DATA_DIR, DATA_DIR)
    print("Done.")
else:
    print("Local dataset copy already exists, skipping.")

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS_HEAD = 15
EPOCHS_FINETUNE = 15
LR_HEAD = 1e-3
LR_FINETUNE = 1e-5
FINETUNE_UNFREEZE_LAST_N = 30

MODEL_CONFIGS = {
    "VGG16": dict(app=vgg16.VGG16, preprocess=vgg16.preprocess_input),
    "ResNet50": dict(app=resnet50.ResNet50, preprocess=resnet50.preprocess_input),
    "DenseNet121": dict(app=densenet.DenseNet121, preprocess=densenet.preprocess_input),
    "EfficientNetB0": dict(app=efficientnet.EfficientNetB0, preprocess=efficientnet.preprocess_input),
    "MobileNetV2": dict(app=mobilenet_v2.MobileNetV2, preprocess=mobilenet_v2.preprocess_input),
}
```

Cell 3 — Data pipeline

```python
def make_datasets():
    train_ds = tf.keras.utils.image_dataset_from_directory(
        TRAIN_DIR, image_size=IMG_SIZE, batch_size=BATCH_SIZE,
        label_mode='binary', shuffle=True, seed=SEED)
    val_ds = tf.keras.utils.image_dataset_from_directory(
        VAL_DIR, image_size=IMG_SIZE, batch_size=BATCH_SIZE,
        label_mode='binary', shuffle=False)
    test_ds = tf.keras.utils.image_dataset_from_directory(
        TEST_DIR, image_size=IMG_SIZE, batch_size=BATCH_SIZE,
        label_mode='binary', shuffle=False)

    class_names = train_ds.class_names
    print("Class order:", class_names)

    train_ds = train_ds.cache()
    val_ds = val_ds.cache()
    test_ds = test_ds.cache()

    aug = tf.keras.Sequential([
        layers.RandomFlip("horizontal"),
        layers.RandomRotation(0.05),
        layers.RandomZoom(0.1),
    ])
    train_ds = train_ds.map(lambda x, y: (aug(x, training=True), y),
                             num_parallel_calls=tf.data.AUTOTUNE)

    AUTOTUNE = tf.data.AUTOTUNE
    train_ds = train_ds.prefetch(AUTOTUNE)
    val_ds = val_ds.prefetch(AUTOTUNE)
    test_ds_eval = test_ds.prefetch(AUTOTUNE)
    return train_ds, val_ds, test_ds_eval, class_names

def compute_class_weight(train_dir, class_names):
    counts = {c: len(os.listdir(os.path.join(train_dir, c))) for c in class_names}
    total = sum(counts.values())
    n_classes = len(class_names)
    weights = {i: total / (n_classes * counts[c]) for i, c in enumerate(class_names)}
    print("Class counts:", counts, " -> class_weight:", weights)
    return weights
```

Cell 4 — Model builder

```python
def build_model(app_fn, preprocess_fn):
    inputs = tf.keras.Input(shape=IMG_SIZE + (3,))
    x = layers.Lambda(preprocess_fn)(inputs)
    base = app_fn(include_top=False, weights='imagenet', input_tensor=x)
    base.trainable = False
    x = layers.GlobalAveragePooling2D()(base.output)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(128, activation='relu')(x)
    x = layers.Dropout(0.2)(x)
    outputs = layers.Dense(1, activation='sigmoid')(x)
    model = Model(inputs, outputs)
    return model, base

def unfreeze_top_layers(base, n):
    base.trainable = True
    for layer in base.layers[:-n]:
        layer.trainable = False
    for layer in base.layers[-n:]:
        if isinstance(layer, layers.BatchNormalization):
            layer.trainable = False

def get_callbacks(ckpt_path):
    return [
        tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True),
        tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, min_lr=1e-7),
        tf.keras.callbacks.ModelCheckpoint(ckpt_path, monitor='val_loss', save_best_only=True,
                                            save_weights_only=True),
    ]
```

Cell 5 — Evaluation & plotting helpers

```python
def plot_curves(history, out_dir, prefix=""):
    os.makedirs(out_dir, exist_ok=True)
    h = history.history
    plt.figure(figsize=(7, 5))
    plt.plot(h['accuracy'], label='Train Accuracy')
    plt.plot(h['val_accuracy'], label='Val Accuracy')
    plt.title(f'{prefix} Training vs Validation Accuracy')
    plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.grid(alpha=0.3)
    plt.savefig(os.path.join(out_dir, 'accuracy_curve.png'), dpi=150, bbox_inches='tight')
    plt.close()

    plt.figure(figsize=(7, 5))
    plt.plot(h['loss'], label='Train Loss')
    plt.plot(h['val_loss'], label='Val Loss')
    plt.title(f'{prefix} Training vs Validation Loss')
    plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.grid(alpha=0.3)
    plt.savefig(os.path.join(out_dir, 'loss_curve.png'), dpi=150, bbox_inches='tight')
    plt.close()

def evaluate_and_document(model, test_ds, class_names, out_dir, model_name, phase):
    os.makedirs(out_dir, exist_ok=True)
    y_true, y_prob = [], []
    for x, y in test_ds:
        y_true.extend(y.numpy().flatten().tolist())
        y_prob.extend(model.predict(x, verbose=0).flatten().tolist())
    y_true = np.array(y_true); y_prob = np.array(y_prob)
    y_pred = (y_prob >= 0.5).astype(int)

    report_dict = classification_report(y_true, y_pred, target_names=class_names, output_dict=True)
    report_df = pd.DataFrame(report_dict).transpose()
    report_df.to_csv(os.path.join(out_dir, 'classification_report.csv'))
    print(f"\n[{model_name} | {phase}] Classification report:\n",
          classification_report(y_true, y_pred, target_names=class_names))

    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(5, 4))
    plt.imshow(cm, cmap='Blues')
    plt.title(f'{model_name} ({phase}) - Confusion Matrix')
    plt.colorbar()
    ticks = np.arange(len(class_names))
    plt.xticks(ticks, class_names); plt.yticks(ticks, class_names)
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, cm[i, j], ha='center', va='center',
                      color='white' if cm[i, j] > cm.max() / 2 else 'black')
    plt.xlabel('Predicted'); plt.ylabel('True'); plt.tight_layout()
    plt.savefig(os.path.join(out_dir, 'confusion_matrix.png'), dpi=150)
    plt.close()

    fpr, tpr, _ = roc_curve(y_true, y_prob)
    roc_auc = auc(fpr, tpr)
    plt.figure(figsize=(6, 5))
    plt.plot(fpr, tpr, label=f'AUC = {roc_auc:.3f}')
    plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
    plt.title(f'{model_name} ({phase}) - ROC Curve')
    plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate')
    plt.legend(); plt.grid(alpha=0.3)
    plt.savefig(os.path.join(out_dir, 'roc_curve.png'), dpi=150, bbox_inches='tight')
    plt.close()

    prec, rec, _ = precision_recall_curve(y_true, y_prob)
    ap = average_precision_score(y_true, y_prob)
    plt.figure(figsize=(6, 5))
    plt.plot(rec, prec, label=f'AP = {ap:.3f}')
    plt.title(f'{model_name} ({phase}) - Precision-Recall Curve')
    plt.xlabel('Recall'); plt.ylabel('Precision')
    plt.legend(); plt.grid(alpha=0.3)
    plt.savefig(os.path.join(out_dir, 'pr_curve.png'), dpi=150, bbox_inches='tight')
    plt.close()

    metrics = {
        "model": model_name, "phase": phase,
        "accuracy": report_dict["accuracy"],
        "precision": report_dict["weighted avg"]["precision"],
        "recall": report_dict["weighted avg"]["recall"],
        "f1": report_dict["weighted avg"]["f1-score"],
        "roc_auc": roc_auc, "avg_precision": ap,
    }
    with open(os.path.join(out_dir, 'metrics.json'), 'w') as f:
        json.dump(metrics, f, indent=2)
    return metrics
```

Cell 6 — Grad-CAM (XAI)

```python
def find_last_conv_layer(base):
    for layer in reversed(base.layers):
        if isinstance(layer, layers.Conv2D):
            return layer.name
    raise ValueError("No Conv2D layer found in base model")

def make_gradcam_heatmap(img_array, model, base, last_conv_layer_name):
    grad_model = Model(inputs=model.inputs,
                        outputs=[base.get_layer(last_conv_layer_name).output, model.output])
    with tf.GradientTape() as tape:
        conv_out, preds = grad_model(img_array)
        loss = preds[:, 0]
    grads = tape.gradient(loss, conv_out)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    conv_out = conv_out[0]
    heatmap = conv_out @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)
    heatmap = tf.maximum(heatmap, 0) / (tf.reduce_max(heatmap) + 1e-8)
    return heatmap.numpy()

def save_gradcam_overlay(orig_img, heatmap, out_path, title=""):
    import cv2
    heatmap_resized = cv2.resize(heatmap, (orig_img.shape[1], orig_img.shape[0]))
    heatmap_uint8 = np.uint8(255 * heatmap_resized)
    heatmap_color = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)
    heatmap_color = cv2.cvtColor(heatmap_color, cv2.COLOR_BGR2RGB)
    overlay = np.uint8(0.6 * orig_img + 0.4 * heatmap_color)

    fig, axes = plt.subplots(1, 3, figsize=(12, 4))
    axes[0].imshow(orig_img.astype(np.uint8)); axes[0].set_title("Original"); axes[0].axis('off')
    axes[1].imshow(heatmap_resized, cmap='jet'); axes[1].set_title("Grad-CAM"); axes[1].axis('off')
    axes[2].imshow(overlay); axes[2].set_title("Overlay"); axes[2].axis('off')
    fig.suptitle(title)
    plt.tight_layout()
    plt.savefig(out_path, dpi=150, bbox_inches='tight')
    plt.close()

def run_gradcam_samples(model, base, test_ds, class_names, out_dir, model_name, n_correct=3, n_wrong=3):
    os.makedirs(out_dir, exist_ok=True)
    last_conv = find_last_conv_layer(base)

    correct_samples, wrong_samples = [], []
    for x, y in test_ds:
        p = model.predict(x, verbose=0).flatten()
        preds = (p >= 0.5).astype(int)
        labels = y.numpy().flatten().astype(int)
        imgs = x.numpy()
        for i in range(len(labels)):
            if preds[i] == labels[i] and len(correct_samples) < n_correct:
                correct_samples.append((imgs[i], labels[i], preds[i], p[i]))
            elif preds[i] != labels[i] and len(wrong_samples) < n_wrong:
                wrong_samples.append((imgs[i], labels[i], preds[i], p[i]))
        if len(correct_samples) >= n_correct and len(wrong_samples) >= n_wrong:
            break

    def process(samples, tag):
        for i, (img, true_l, pred_l, prob) in enumerate(samples):
            img_batch = np.expand_dims(img, axis=0)
            heatmap = make_gradcam_heatmap(img_batch, model, base, last_conv)
            title = f"{model_name} | true={class_names[true_l]} pred={class_names[pred_l]} (p={prob:.2f})"
            save_gradcam_overlay(img, heatmap, os.path.join(out_dir, f"{tag}_{i}.png"), title=title)

    process(correct_samples, "correct")
    process(wrong_samples, "misclassified")
    print(f"[{model_name}] Grad-CAM saved ({len(correct_samples)} correct, {len(wrong_samples)} misclassified)")
```

Cell 7 — Main training loop

```python
def is_done(phase_dir):
    return os.path.exists(os.path.join(phase_dir, "metrics.json"))

def load_metrics(phase_dir):
    with open(os.path.join(phase_dir, "metrics.json")) as f:
        return json.load(f)

def make_datasets_cached():
    train_ds, val_ds, test_ds, class_names = make_datasets_raw()
    return train_ds, val_ds, test_ds, class_names

# Redefine make_datasets to cache to local disk instead of RAM
def make_datasets():
    train_ds = tf.keras.utils.image_dataset_from_directory(
        TRAIN_DIR, image_size=IMG_SIZE, batch_size=BATCH_SIZE,
        label_mode='binary', shuffle=True, seed=SEED)
    val_ds = tf.keras.utils.image_dataset_from_directory(
        VAL_DIR, image_size=IMG_SIZE, batch_size=BATCH_SIZE,
        label_mode='binary', shuffle=False)
    test_ds = tf.keras.utils.image_dataset_from_directory(
        TEST_DIR, image_size=IMG_SIZE, batch_size=BATCH_SIZE,
        label_mode='binary', shuffle=False)

    class_names = train_ds.class_names
    print("Class order:", class_names)

    os.makedirs("/content/tfcache", exist_ok=True)
    train_ds = train_ds.cache("/content/tfcache/train")
    val_ds = val_ds.cache("/content/tfcache/val")
    test_ds = test_ds.cache("/content/tfcache/test")

    aug = tf.keras.Sequential([
        layers.RandomFlip("horizontal"),
        layers.RandomRotation(0.05),
        layers.RandomZoom(0.1),
    ])
    train_ds = train_ds.map(lambda x, y: (aug(x, training=True), y),
                             num_parallel_calls=tf.data.AUTOTUNE)

    AUTOTUNE = tf.data.AUTOTUNE
    train_ds = train_ds.prefetch(AUTOTUNE)
    val_ds = val_ds.prefetch(AUTOTUNE)
    test_ds_eval = test_ds.prefetch(AUTOTUNE)
    return train_ds, val_ds, test_ds_eval, class_names


def train_all_models():
    train_ds, val_ds, test_ds, class_names = make_datasets()
    class_weight = compute_class_weight(TRAIN_DIR, class_names)
    all_metrics = []

    for model_name, cfg in MODEL_CONFIGS.items():
        model_dir = os.path.join(RESULTS_DIR, model_name)
        phase1_dir = os.path.join(model_dir, "phase1_no_finetune")
        phase2_dir = os.path.join(model_dir, "phase2_finetuned")
        os.makedirs(phase1_dir, exist_ok=True)
        os.makedirs(phase2_dir, exist_ok=True)

        if is_done(phase1_dir) and is_done(phase2_dir):
            print(f"\n{model_name}: already fully completed, skipping.")
            all_metrics.append(load_metrics(phase1_dir))
            all_metrics.append(load_metrics(phase2_dir))
            continue

        print("\n" + "=" * 70); print(f"MODEL: {model_name}"); print("=" * 70)

        tf.keras.backend.clear_session()
        model, base = build_model(cfg["app"], cfg["preprocess"])
        ckpt1 = os.path.join(phase1_dir, "best.weights.h5")
        ckpt2 = os.path.join(phase2_dir, "best.weights.h5")

        # ---- Phase 1 ----
        model.compile(optimizer=tf.keras.optimizers.Adam(LR_HEAD),
                      loss='binary_crossentropy', metrics=['accuracy'])
        if is_done(phase1_dir):
            print(f"{model_name}: phase 1 already done, loading checkpoint.")
            model.load_weights(ckpt1)
            metrics1 = load_metrics(phase1_dir)
        else:
            history1 = model.fit(train_ds, validation_data=val_ds, epochs=EPOCHS_HEAD,
                                  class_weight=class_weight, callbacks=get_callbacks(ckpt1))
            plot_curves(history1, phase1_dir, prefix=f"{model_name} (No Fine-tune)")
            metrics1 = evaluate_and_document(model, test_ds, class_names, phase1_dir, model_name, "no_finetune")
            del history1
        all_metrics.append(metrics1)

        # ---- Phase 2 ----
        unfreeze_top_layers(base, FINETUNE_UNFREEZE_LAST_N)
        model.compile(optimizer=tf.keras.optimizers.Adam(LR_FINETUNE),
                      loss='binary_crossentropy', metrics=['accuracy'])
        if is_done(phase2_dir):
            print(f"{model_name}: phase 2 already done, loading checkpoint.")
            model.load_weights(ckpt2)
            metrics2 = load_metrics(phase2_dir)
        else:
            history2 = model.fit(train_ds, validation_data=val_ds, epochs=EPOCHS_FINETUNE,
                                  class_weight=class_weight, callbacks=get_callbacks(ckpt2))
            plot_curves(history2, phase2_dir, prefix=f"{model_name} (Fine-tuned)")
            metrics2 = evaluate_and_document(model, test_ds, class_names, phase2_dir, model_name, "finetuned")
            del history2
        all_metrics.append(metrics2)

        gradcam_dir = os.path.join(model_dir, "gradcam")
        if not os.path.exists(os.path.join(gradcam_dir, "correct_0.png")):
            try:
                run_gradcam_samples(model, base, test_ds, class_names, gradcam_dir, model_name)
            except Exception as e:
                print(f"Grad-CAM failed for {model_name}: {e}")

        final_path = os.path.join(model_dir, f"{model_name}_final.keras")
        if not os.path.exists(final_path):
            model.save(final_path)

        del model, base
        gc.collect()
        tf.keras.backend.clear_session()

    comp_df = pd.DataFrame(all_metrics)
    comp_df.to_csv(os.path.join(RESULTS_DIR, "model_comparison.csv"), index=False)
    print("\nFinal comparison table:\n", comp_df)
    plot_comparison_chart(comp_df, RESULTS_DIR)
    return comp_df
```

Cell 8 — Comparison chart

```python
def plot_comparison_chart(comp_df, out_dir):
    finetuned = comp_df[comp_df.phase == "finetuned"].set_index("model")
    no_ft = comp_df[comp_df.phase == "no_finetune"].set_index("model")
    metrics_to_plot = ["accuracy", "precision", "recall", "f1", "roc_auc"]
    x = np.arange(len(finetuned.index)); width = 0.35

    for metric in metrics_to_plot:
        plt.figure(figsize=(8, 5))
        plt.bar(x - width/2, no_ft[metric], width, label="No Fine-tune")
        plt.bar(x + width/2, finetuned[metric], width, label="Fine-tuned")
        plt.xticks(x, finetuned.index, rotation=20)
        plt.ylabel(metric.replace("_", " ").title())
        plt.title(f"Model Comparison - {metric.replace('_', ' ').title()}")
        plt.ylim(0, 1.05); plt.legend(); plt.grid(alpha=0.3, axis='y')
        plt.tight_layout()
        plt.savefig(os.path.join(out_dir, f"comparison_{metric}.png"), dpi=150)
        plt.close()
    print(f"Comparison charts saved to {out_dir}")
```

```python
def show_phase_results(model_name, phase_label, phase_dir):
    if not os.path.exists(os.path.join(phase_dir, "metrics.json")):
        display(Markdown(f"### {model_name} — {phase_label}: *not completed yet*"))
        return

    display(Markdown(f"## {model_name} — {phase_label}"))

    report_path = os.path.join(phase_dir, "classification_report.csv")
    if os.path.exists(report_path):
        df = pd.read_csv(report_path, index_col=0)
        display(Markdown("**Classification Report:**"))
        display(df.round(4))

    image_files = [
        ("accuracy_curve.png", "Training vs Validation Accuracy"),
        ("loss_curve.png", "Training vs Validation Loss"),
        ("confusion_matrix.png", "Confusion Matrix"),
        ("roc_curve.png", "ROC Curve"),
        ("pr_curve.png", "Precision-Recall Curve"),
    ]
    for fname, title in image_files:
        img_path = os.path.join(phase_dir, fname)
        if os.path.exists(img_path):
            display(Markdown(f"**{title}:**"))
            display(IPImage(filename=img_path))
    print("\n")
```

```python
import matplotlib.pyplot as plt
import random

def show_dataset_overview(split_dir, title="Dataset Overview"):
    fig, axes = plt.subplots(2, 4, figsize=(14, 7))
    for row, cls in enumerate(["NORMAL", "PNEUMONIA"]):
        cls_dir = os.path.join(split_dir, cls)
        sample_files = random.sample(os.listdir(cls_dir), 4)
        for col, fname in enumerate(sample_files):
            img = plt.imread(os.path.join(cls_dir, fname))
            axes[row, col].imshow(img, cmap='gray')
            axes[row, col].set_title(cls if col == 0 else "")
            axes[row, col].axis('off')
    fig.suptitle(title)
    plt.tight_layout()
    plt.savefig(os.path.join(RESULTS_DIR, "dataset_sample_grid.png"), dpi=150, bbox_inches='tight')
    plt.show()

show_dataset_overview(os.path.join("/content/drive/MyDrive/Chest X-ray Pneumonia/Final_Dataset_Split", "train"))

# Class distribution across splits
counts = []
for split in ["train", "val", "test"]:
    for cls in ["NORMAL", "PNEUMONIA"]:
        d = f"/content/drive/MyDrive/Chest X-ray Pneumonia/Final_Dataset_Split/{split}/{cls}"
        counts.append({"split": split, "class": cls, "count": len(os.listdir(d))})
dist_df = pd.DataFrame(counts)
display(dist_df.pivot(index="split", columns="class", values="count"))

dist_df.pivot(index="split", columns="class", values="count").plot(kind="bar", figsize=(7, 5))
plt.title("Class Distribution Across Splits")
plt.ylabel("Image Count")
plt.savefig(os.path.join(RESULTS_DIR, "class_distribution.png"), dpi=150, bbox_inches='tight')
plt.show()
```

Cell R7 — Sensitivity & Specificity table (all models, both phases)

```python
def get_sens_spec(report_path):
    df = pd.read_csv(report_path, index_col=0)
    sensitivity = df.loc["PNEUMONIA", "recall"]  # recall on positive class
    specificity = df.loc["NORMAL", "recall"]     # recall on negative class
    return sensitivity, specificity

rows = []
for model_name in MODEL_NAMES:
    for phase, phase_dir_name in [("no_finetune", "phase1_no_finetune"), ("finetuned", "phase2_finetuned")]:
        report_path = os.path.join(RESULTS_DIR, model_name, phase_dir_name, "classification_report.csv")
        if os.path.exists(report_path):
            sens, spec = get_sens_spec(report_path)
            rows.append({"model": model_name, "phase": phase, "sensitivity": sens, "specificity": spec})

sens_spec_df = pd.DataFrame(rows)
display(Markdown("## Sensitivity & Specificity — All Models"))
display(sens_spec_df.round(4))
sens_spec_df.to_csv(os.path.join(RESULTS_DIR, "sensitivity_specificity.csv"), index=False)
```

Cell R8 — Model complexity table

```python
# Requires build_model() from your training notebook (Cell 4) to be defined in this session.
# If running in a fresh notebook, re-run Cell 4's build_model function first.

complexity_rows = []
for model_name, cfg in MODEL_CONFIGS.items():
    tf.keras.backend.clear_session()
    model, base = build_model(cfg["app"], cfg["preprocess"])
    total_params = model.count_params()
    trainable_params = sum([tf.size(w).numpy() for w in model.trainable_weights])
    complexity_rows.append({
        "model": model_name,
        "total_params": total_params,
        "trainable_params_frozen_phase": trainable_params,
    })
    del model, base

complexity_df = pd.DataFrame(complexity_rows)
display(Markdown("## Model Complexity"))
display(complexity_df)
complexity_df.to_csv(os.path.join(RESULTS_DIR, "model_complexity.csv"), index=False)
```
