Preprocessing helpers

```python
def list_images(folder):
    files = []
    for ext in VALID_EXTS:
        files.extend(glob.glob(os.path.join(folder, f"*{ext}")))
        files.extend(glob.glob(os.path.join(folder, f"*{ext.upper()}")))
    return sorted(files)

def autocrop_black_border(img_rgb, thresh=15, min_area_ratio=0.4):
    gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
    mask = (gray > thresh).astype(np.uint8)
    coords = cv2.findNonZero(mask)
    if coords is None:
        return img_rgb
    x, y, w, h = cv2.boundingRect(coords)
    area_ratio = (w * h) / (gray.shape[0] * gray.shape[1])
    if area_ratio < min_area_ratio or area_ratio > 0.98:
        return img_rgb
    return img_rgb[y:y + h, x:x + w]

def standardize_contrast(img_rgb):
    gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    gray_eq = clahe.apply(gray)
    return cv2.cvtColor(gray_eq, cv2.COLOR_GRAY2RGB)

def load_and_preprocess(path, size=IMG_SIZE, crop_border=False):
    img = cv2.imread(path, cv2.IMREAD_COLOR)
    if img is None:
        return None
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    if crop_border:
        img = autocrop_black_border(img)
    img = cv2.resize(img, size, interpolation=cv2.INTER_AREA)
    img = standardize_contrast(img)
    return img

def save_image(img_rgb, path):
    img_bgr = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR)
    cv2.imwrite(path, img_bgr)

def clear_and_make_dir(path):
    if os.path.exists(path):
        shutil.rmtree(path)
    os.makedirs(path, exist_ok=True)

def load_folder(folder, crop_border=False, desc=""):
    paths = list_images(folder)
    imgs = []
    for p in tqdm(paths, desc=desc):
        img = load_and_preprocess(p, crop_border=crop_border)
        if img is not None:
            imgs.append(img)
        else:
            print(f"  Skipping unreadable file: {p}")
    return imgs
```

Augmentation (used only on clean/primary images)

```python
def augment_image(img):
    h, w = img.shape[:2]
    choice = random.choice(["flip", "rotate", "zoom", "brightness", "shift", "noise"])
    if choice == "flip":
        out = cv2.flip(img, 1)
    elif choice == "rotate":
        angle = random.uniform(-15, 15)
        M = cv2.getRotationMatrix2D((w / 2, h / 2), angle, 1.0)
        out = cv2.warpAffine(img, M, (w, h), borderMode=cv2.BORDER_REFLECT)
    elif choice == "zoom":
        zf = random.uniform(1.05, 1.25)
        zh, zw = int(h / zf), int(w / zf)
        y0 = random.randint(0, h - zh)
        x0 = random.randint(0, w - zw)
        out = cv2.resize(img[y0:y0 + zh, x0:x0 + zw], (w, h), interpolation=cv2.INTER_AREA)
    elif choice == "brightness":
        factor = random.uniform(0.8, 1.2)
        out = np.clip(img.astype(np.float32) * factor, 0, 255).astype(np.uint8)
    elif choice == "shift":
        tx = random.randint(-int(0.1 * w), int(0.1 * w))
        ty = random.randint(-int(0.1 * h), int(0.1 * h))
        M = np.float32([[1, 0, tx], [0, 1, ty]])
        out = cv2.warpAffine(img, M, (w, h), borderMode=cv2.BORDER_REFLECT)
    else:
        noise = np.random.normal(0, 6, img.shape).astype(np.int16)
        out = np.clip(img.astype(np.int16) + noise, 0, 255).astype(np.uint8)
    return out

def fill_with_augmentation(pool_imgs, n_needed):
    out = []
    for _ in tqdm(range(n_needed), desc="Augmenting"):
        base = random.choice(pool_imgs)
        out.append(augment_image(base))
    return out

def save_all(imgs, dst_folder, prefix, start_idx=1):
    idx = start_idx
    for img in imgs:
        save_image(img, os.path.join(dst_folder, f"{prefix}_{idx:06d}.jpg"))
        idx += 1
    return idx - start_idx
```

Data pipeline

```python
def make_datasets():
    train_ds = tf.keras.utils.image_dataset_from_directory(
        TRAIN_DIR, image_size=IMG_SIZE, batch_size=BATCH_SIZE,
        label_mode='binary', shuffle=True, seed=SEED)
    val_ds = tf.keras.utils.image_dataset_from_directory(
        VAL_DIR, image_size=IMG_SIZE, batch_size=BATCH_SIZE,
        label_mode='binary', shuffle=False)
    test_ds = tf.keras.utils.image_dataset_from_directory(
        TEST_DIR, image_size=IMG_SIZE, batch_size=BATCH_SIZE,
        label_mode='binary', shuffle=False)

    class_names = train_ds.class_names
    print("Class order:", class_names)

    aug = tf.keras.Sequential([
        layers.RandomFlip("horizontal"),
        layers.RandomRotation(0.05),
        layers.RandomZoom(0.1),
    ])
    train_ds = train_ds.map(lambda x, y: (aug(x, training=True), y),
                             num_parallel_calls=tf.data.AUTOTUNE)

    AUTOTUNE = tf.data.AUTOTUNE
    train_ds = train_ds.prefetch(AUTOTUNE)
    val_ds = val_ds.prefetch(AUTOTUNE)
    test_ds_eval = test_ds.prefetch(AUTOTUNE)
    return train_ds, val_ds, test_ds_eval, class_names

def compute_class_weight(train_dir, class_names):
    counts = {c: len(os.listdir(os.path.join(train_dir, c))) for c in class_names}
    total = sum(counts.values())
    n_classes = len(class_names)
    weights = {i: total / (n_classes * counts[c]) for i, c in enumerate(class_names)}
    print("Class counts:", counts, " -> class_weight:", weights)
    return weights
```

Model builder

```python
def build_model(app_fn, preprocess_fn):
    inputs = tf.keras.Input(shape=IMG_SIZE + (3,))
    x = layers.Lambda(preprocess_fn)(inputs)
    base = app_fn(include_top=False, weights='imagenet', input_tensor=x)
    base.trainable = False
    x = layers.GlobalAveragePooling2D()(base.output)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(128, activation='relu')(x)
    x = layers.Dropout(0.2)(x)
    outputs = layers.Dense(1, activation='sigmoid')(x)
    model = Model(inputs, outputs)
    return model, base

def unfreeze_top_layers(base, n):
    base.trainable = True
    for layer in base.layers[:-n]:
        layer.trainable = False
    for layer in base.layers[-n:]:
        if isinstance(layer, layers.BatchNormalization):
            layer.trainable = False

def get_callbacks(ckpt_path):
    return [
        tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True),
        tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, min_lr=1e-7),
        tf.keras.callbacks.ModelCheckpoint(ckpt_path, monitor='val_loss', save_best_only=True),
    ]
```

Evaluation & plotting helpers

```python
def plot_curves(history, out_dir, prefix=""):
    os.makedirs(out_dir, exist_ok=True)
    h = history.history
    plt.figure(figsize=(7, 5))
    plt.plot(h['accuracy'], label='Train Accuracy')
    plt.plot(h['val_accuracy'], label='Val Accuracy')
    plt.title(f'{prefix} Training vs Validation Accuracy')
    plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.grid(alpha=0.3)
    plt.savefig(os.path.join(out_dir, 'accuracy_curve.png'), dpi=150, bbox_inches='tight')
    plt.close()

    plt.figure(figsize=(7, 5))
    plt.plot(h['loss'], label='Train Loss')
    plt.plot(h['val_loss'], label='Val Loss')
    plt.title(f'{prefix} Training vs Validation Loss')
    plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.grid(alpha=0.3)
    plt.savefig(os.path.join(out_dir, 'loss_curve.png'), dpi=150, bbox_inches='tight')
    plt.close()

def evaluate_and_document(model, test_ds, class_names, out_dir, model_name, phase):
    os.makedirs(out_dir, exist_ok=True)
    y_true, y_prob = [], []
    for x, y in test_ds:
        y_true.extend(y.numpy().flatten().tolist())
        y_prob.extend(model.predict(x, verbose=0).flatten().tolist())
    y_true = np.array(y_true); y_prob = np.array(y_prob)
    y_pred = (y_prob >= 0.5).astype(int)

    report_dict = classification_report(y_true, y_pred, target_names=class_names, output_dict=True)
    report_df = pd.DataFrame(report_dict).transpose()
    report_df.to_csv(os.path.join(out_dir, 'classification_report.csv'))
    print(f"\n[{model_name} | {phase}] Classification report:\n",
          classification_report(y_true, y_pred, target_names=class_names))

    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(5, 4))
    plt.imshow(cm, cmap='Blues')
    plt.title(f'{model_name} ({phase}) - Confusion Matrix')
    plt.colorbar()
    ticks = np.arange(len(class_names))
    plt.xticks(ticks, class_names); plt.yticks(ticks, class_names)
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, cm[i, j], ha='center', va='center',
                      color='white' if cm[i, j] > cm.max() / 2 else 'black')
    plt.xlabel('Predicted'); plt.ylabel('True'); plt.tight_layout()
    plt.savefig(os.path.join(out_dir, 'confusion_matrix.png'), dpi=150)
    plt.close()

    fpr, tpr, _ = roc_curve(y_true, y_prob)
    roc_auc = auc(fpr, tpr)
    plt.figure(figsize=(6, 5))
    plt.plot(fpr, tpr, label=f'AUC = {roc_auc:.3f}')
    plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
    plt.title(f'{model_name} ({phase}) - ROC Curve')
    plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate')
    plt.legend(); plt.grid(alpha=0.3)
    plt.savefig(os.path.join(out_dir, 'roc_curve.png'), dpi=150, bbox_inches='tight')
    plt.close()

    prec, rec, _ = precision_recall_curve(y_true, y_prob)
    ap = average_precision_score(y_true, y_prob)
    plt.figure(figsize=(6, 5))
    plt.plot(rec, prec, label=f'AP = {ap:.3f}')
    plt.title(f'{model_name} ({phase}) - Precision-Recall Curve')
    plt.xlabel('Recall'); plt.ylabel('Precision')
    plt.legend(); plt.grid(alpha=0.3)
    plt.savefig(os.path.join(out_dir, 'pr_curve.png'), dpi=150, bbox_inches='tight')
    plt.close()

    metrics = {
        "model": model_name, "phase": phase,
        "accuracy": report_dict["accuracy"],
        "precision": report_dict["weighted avg"]["precision"],
        "recall": report_dict["weighted avg"]["recall"],
        "f1": report_dict["weighted avg"]["f1-score"],
        "roc_auc": roc_auc, "avg_precision": ap,
    }
    with open(os.path.join(out_dir, 'metrics.json'), 'w') as f:
        json.dump(metrics, f, indent=2)
    return metrics
```

Grad-CAM (item 5: XAI)

```python
def find_last_conv_layer(base):
    for layer in reversed(base.layers):
        if isinstance(layer, layers.Conv2D):
            return layer.name
    raise ValueError("No Conv2D layer found in base model")

def make_gradcam_heatmap(img_array, model, base, last_conv_layer_name):
    grad_model = Model(inputs=model.inputs,
                        outputs=[base.get_layer(last_conv_layer_name).output, model.output])
    with tf.GradientTape() as tape:
        conv_out, preds = grad_model(img_array)
        loss = preds[:, 0]
    grads = tape.gradient(loss, conv_out)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    conv_out = conv_out[0]
    heatmap = conv_out @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)
    heatmap = tf.maximum(heatmap, 0) / (tf.reduce_max(heatmap) + 1e-8)
    return heatmap.numpy()

def save_gradcam_overlay(orig_img, heatmap, out_path, title=""):
    import cv2
    heatmap_resized = cv2.resize(heatmap, (orig_img.shape[1], orig_img.shape[0]))
    heatmap_uint8 = np.uint8(255 * heatmap_resized)
    heatmap_color = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)
    heatmap_color = cv2.cvtColor(heatmap_color, cv2.COLOR_BGR2RGB)
    overlay = np.uint8(0.6 * orig_img + 0.4 * heatmap_color)

    fig, axes = plt.subplots(1, 3, figsize=(12, 4))
    axes[0].imshow(orig_img.astype(np.uint8)); axes[0].set_title("Original"); axes[0].axis('off')
    axes[1].imshow(heatmap_resized, cmap='jet'); axes[1].set_title("Grad-CAM"); axes[1].axis('off')
    axes[2].imshow(overlay); axes[2].set_title("Overlay"); axes[2].axis('off')
    fig.suptitle(title)
    plt.tight_layout()
    plt.savefig(out_path, dpi=150, bbox_inches='tight')
    plt.close()

def run_gradcam_samples(model, base, test_ds, class_names, out_dir, model_name, n_correct=3, n_wrong=3):
    os.makedirs(out_dir, exist_ok=True)
    last_conv = find_last_conv_layer(base)

    imgs, labels, probs = [], [], []
    for x, y in test_ds:
        p = model.predict(x, verbose=0).flatten()
        imgs.append(x.numpy()); labels.append(y.numpy().flatten()); probs.append(p)
    imgs = np.concatenate(imgs); labels = np.concatenate(labels); probs = np.concatenate(probs)
    preds = (probs >= 0.5).astype(int)

    correct_idx = np.where(preds == labels)[0]
    wrong_idx = np.where(preds != labels)[0]
    random.shuffle(correct_idx.tolist()); random.shuffle(wrong_idx.tolist())

    def process(idx_list, tag, k):
        for i, idx in enumerate(idx_list[:k]):
            img = imgs[idx]
            img_batch = np.expand_dims(img, axis=0)
            heatmap = make_gradcam_heatmap(img_batch, model, base, last_conv)
            true_c = class_names[int(labels[idx])]
            pred_c = class_names[int(preds[idx])]
            title = f"{model_name} | true={true_c} pred={pred_c} (p={probs[idx]:.2f})"
            save_gradcam_overlay(img, heatmap, os.path.join(out_dir, f"{tag}_{i}.png"), title=title)

    process(correct_idx, "correct", n_correct)
    process(wrong_idx, "misclassified", n_wrong)
    print(f"[{model_name}] Grad-CAM samples saved to {out_dir}")
```

Main training loop (this runs everything)

```python
def train_all_models():
    train_ds, val_ds, test_ds, class_names = make_datasets()
    class_weight = compute_class_weight(TRAIN_DIR, class_names)
    all_metrics = []

    for model_name, cfg in MODEL_CONFIGS.items():
        print("\n" + "=" * 70); print(f"MODEL: {model_name}"); print("=" * 70)
        model_dir = os.path.join(RESULTS_DIR, model_name)
        os.makedirs(model_dir, exist_ok=True)

        tf.keras.backend.clear_session()
        model, base = build_model(cfg["app"], cfg["preprocess"])

        # Phase 1: frozen base
        model.compile(optimizer=tf.keras.optimizers.Adam(LR_HEAD),
                      loss='binary_crossentropy', metrics=['accuracy'])
        phase1_dir = os.path.join(model_dir, "phase1_no_finetune")
        os.makedirs(phase1_dir, exist_ok=True)
        history1 = model.fit(train_ds, validation_data=val_ds, epochs=EPOCHS_HEAD,
                              class_weight=class_weight,
                              callbacks=get_callbacks(os.path.join(phase1_dir, "best.weights.h5")))
        plot_curves(history1, phase1_dir, prefix=f"{model_name} (No Fine-tune)")
        metrics1 = evaluate_and_document(model, test_ds, class_names, phase1_dir, model_name, "no_finetune")
        all_metrics.append(metrics1)

        # Phase 2: fine-tune
        unfreeze_top_layers(base, FINETUNE_UNFREEZE_LAST_N)
        model.compile(optimizer=tf.keras.optimizers.Adam(LR_FINETUNE),
                      loss='binary_crossentropy', metrics=['accuracy'])
        phase2_dir = os.path.join(model_dir, "phase2_finetuned")
        os.makedirs(phase2_dir, exist_ok=True)
        history2 = model.fit(train_ds, validation_data=val_ds, epochs=EPOCHS_FINETUNE,
                              class_weight=class_weight,
                              callbacks=get_callbacks(os.path.join(phase2_dir, "best.weights.h5")))
        plot_curves(history2, phase2_dir, prefix=f"{model_name} (Fine-tuned)")
        metrics2 = evaluate_and_document(model, test_ds, class_names, phase2_dir, model_name, "finetuned")
        all_metrics.append(metrics2)

        gradcam_dir = os.path.join(model_dir, "gradcam")
        try:
            run_gradcam_samples(model, base, test_ds, class_names, gradcam_dir, model_name)
        except Exception as e:
            print(f"Grad-CAM failed for {model_name}: {e}")

        model.save(os.path.join(model_dir, f"{model_name}_final.keras"))
        del model, base, history1, history2
        gc.collect()

    comp_df = pd.DataFrame(all_metrics)
    comp_df.to_csv(os.path.join(RESULTS_DIR, "model_comparison.csv"), index=False)
    print("\nFinal comparison table:\n", comp_df)
    plot_comparison_chart(comp_df, RESULTS_DIR)
    return comp_df
```

```python
def plot_comparison_chart(comp_df, out_dir):
    finetuned = comp_df[comp_df.phase == "finetuned"].set_index("model")
    no_ft = comp_df[comp_df.phase == "no_finetune"].set_index("model")
    metrics_to_plot = ["accuracy", "precision", "recall", "f1", "roc_auc"]
    x = np.arange(len(finetuned.index)); width = 0.35

    for metric in metrics_to_plot:
        plt.figure(figsize=(8, 5))
        plt.bar(x - width/2, no_ft[metric], width, label="No Fine-tune")
        plt.bar(x + width/2, finetuned[metric], width, label="Fine-tuned")
        plt.xticks(x, finetuned.index, rotation=20)
        plt.ylabel(metric.replace("_", " ").title())
        plt.title(f"Model Comparison - {metric.replace('_', ' ').title()}")
        plt.ylim(0, 1.05); plt.legend(); plt.grid(alpha=0.3, axis='y')
        plt.tight_layout()
        plt.savefig(os.path.join(out_dir, f"comparison_{metric}.png"), dpi=150)
        plt.close()
    print(f"Comparison charts saved to {out_dir}")
```
